```python
from rdkit import Chem
import pandas as pd

def smiles_to_mol(smiles):
    if pd.isna(smiles):
        return None
    mol = Chem.MolFromSmiles(str(smiles))
    return mol

# Use 'df_clean' instead of 'df'
df_clean["mol"] = df_clean["canonical_smiles"].apply(smiles_to_mol)

print("Total molecules:", len(df_clean))
print("Failed conversions:", df_clean["mol"].isna().sum())
```

```python
from rdkit.Chem import Descriptors

descriptor_functions = [
    (name, func)
    for name, func in Descriptors.descList
]

descriptor_names = [name for name, func in descriptor_functions]

print("Number of descriptors:", len(descriptor_names))
```

```python
import numpy as np
import pandas as pd

descriptor_data = []

# Corrected the NameError by using df_clean instead of df
for mol in df_clean["mol"]:
    values = []

    # Check if the molecule is valid (not None) before extracting descriptors
    if mol is None:
        # Fill all descriptor columns with NaN for this row
        values = [np.nan] * len(descriptor_functions)
    else:
        for name, func in descriptor_functions:
            try:
                values.append(func(mol))
            except:
                values.append(np.nan)

    descriptor_data.append(values)

X_desc = pd.DataFrame(
    descriptor_data,
    columns=descriptor_names
)

print("Descriptor matrix shape:", X_desc.shape)
```

```python
import numpy as np
from rdkit.Chem import rdFingerprintGenerator

morgan_generator = rdFingerprintGenerator.GetMorganGenerator(
    radius=2,
    fpSize=2048
)

fingerprints = []

# Corrected 'df' to 'df_clean' to fix NameError
for mol in df_clean["mol"]:
    if mol is None:
        # If the molecule is invalid, append a vector of zeros
        fp = np.zeros((2048,), dtype=np.int32)
    else:
        fp = morgan_generator.GetFingerprintAsNumPy(mol)

    fingerprints.append(fp)

X_morgan = np.array(fingerprints)

print("Morgan fingerprint shape:", X_morgan.shape)
```
